```python
duplicates = df[df['Date'].duplicated()]
print(duplicates)
```

```python
# Step 1: Try converting amount column to numeric
df['amount'] = pd.to_numeric(df['amount'], errors='coerce')

# Step 2: Drop rows where amount became NaN (non-numeric ones)
df = df.dropna(subset=['amount'])

# Step 3: Optional - Reset index
df = df.reset_index(drop=True)
```

```python
# Make sure Date is datetime (optional but good practice)
df['Date'] = pd.to_datetime(df['Date'])

# Group by Date and sum the amount
df = df.groupby('Date', as_index=False).agg({'amount': 'sum'})

# Optional: sort the dates
df = df.sort_values('Date').reset_index(drop=True)
```

```python
# 1. Ensure 'Date' column is datetime type
df['Date'] = pd.to_datetime(df['Date'])

# 2. Create the full range of dates from start to end
full_date_range = pd.date_range(start=df['Date'].min(), end=df['Date'].max(), freq='D')

# 3. Find missing dates
missing_dates = full_date_range.difference(df['Date'])

# 4. Print missing dates
print(missing_dates)
```

```python
missing_dates_df = pd.DataFrame(missing_dates, columns=['Missing Dates'])
missing_dates_df.to_csv('missing_dates.csv', index=False)
```
